# Motorcycle Helmet Detection using YOLOv8

This notebook trains a YOLOv8 model for real-time helmet detection in motorcycle safety applications.

**Key Features:**
- Real-time detection
- High accuracy with YOLOv8 architecture
- Production-ready inference
- Easy deployment

## 1. Setup and Imports

```python
!pip install -q ultralytics roboflow opencv-python

from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
from pathlib import Path
```

## 2. Download Dataset from Roboflow

```python
from roboflow import Roboflow

# Initialize Roboflow
rf = Roboflow(api_key='YOUR_API_KEY')
project = rf.workspace().project('helmet-detection')
dataset = project.versions(1).download('yolov8')
```

## 3. Train YOLOv8 Model

```python
# Load YOLOv8 model
model = YOLO('yolov8m.pt')

# Train
results = model.train(
    data='data.yaml',
    epochs=100,
    imgsz=640,
    device=0,
    patience=20
)
```

## 4. Evaluate Model

```python
# Validate
metrics = model.val()
print(f'mAP50: {metrics.box.map50}')
print(f'mAP50-95: {metrics.box.map}')
```

## 5. Inference on Test Image

```python
# Run inference
results = model.predict(source='test_image.jpg', conf=0.5)

# Display results
for r in results:
    im_array = r.plot()
    plt.figure(figsize=(12, 8))
    plt.imshow(im_array[:, :, ::-1])
    plt.axis('off')
    plt.title('Helmet Detection')
    plt.show()
```

## 6. Export Model

```python
# Export to ONNX for production deployment
path = model.export(format='onnx')
print(f'Model exported to: {path}')
```
